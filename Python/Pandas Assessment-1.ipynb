```python
#Ans14.
numeric = df.select_dtypes(include='number')

corr = numeric.corr()['Close Price'].sort_values(ascending=False)

print(corr)

corr = numeric.corr()['Close Price'].drop('Close Price')

print(corr.sort_values(ascending=False))
#The indicator with the highest correlation is the strongest linear predictor of stock closing price.
"""
Positive correlation (close to +1): As the indicator increases, the stock close price also tends to increase.
Negative correlation (close to -1): As the indicator increases, the stock close price tends to decrease.
Correlation near 0: There is little or no linear relationship.
"""
```
