```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
Sales = pd.read_csv("C:/Users/Taqi Saiyed/Downloads/historical_automobile_sales.csv")
Sales
```

```python
#Ans1.
df8 = Sales.groupby(['Year']).agg({'Automobile_Sales': 'sum'})
df8.reset_index(inplace=True)
df8.head()
```

```python
plt.figure(figsize = (8,5))
plt.plot(df8['Year'], df8['Automobile_Sales'], marker='o', color='pink', linestyle='-', markersize=5)
plt.xlabel('Year')
plt.ylabel('Automobile Sales')
plt.title('Yearwise Automobile Sales')
plt.xlabel('Years')
plt.ylabel('Sales')
plt.grid(True)
plt.show()
```

```python
#Ans2.
# Group by Year and Vehicle Type
vehicle_sales = Sales.groupby(["Year","Vehicle_Type"])["Automobile_Sales"].sum().reset_index()

plt.figure(figsize=(12,6))

for vehicle in vehicle_sales["Vehicle_Type"].unique():
    data = vehicle_sales[vehicle_sales["Vehicle_Type"] == vehicle]
    plt.plot(data["Year"], data["Automobile_Sales"], marker='o', label=vehicle)

plt.title("Sales Trend by Vehicle Type")
plt.xlabel("Year")
plt.ylabel("Automobile Sales")
plt.legend()
plt.grid(True)

plt.show()
```

```python
#Ans3.
plt.figure(figsize=(12,6))

sns.lineplot(
    data=Sales,
    x="Year",
    y="Automobile_Sales",
    hue="Recession",
    style="Vehicle_Type",
    marker="o"
)

plt.title("Vehicle Sales During Recession and Non-Recession")
plt.xlabel("Year")
plt.ylabel("Automobile Sales")

plt.show()
```

```python
#Own Code
df9 = Sales.groupby(['Vehicle_Type','Year','Recession']).agg({'Automobile_Sales':'sum'})
df9.reset_index(inplace=True)
df9
plt.figure(figsize=(12,6))
sns.lineplot(
    data=df9,
    x="Year",
    y="Automobile_Sales",
    hue="Recession",
    style="Vehicle_Type",
    marker="o"
)
plt.show()
```

```python
#Ans4.
# Average sales
s = Sales.groupby(["Recession","Vehicle_Type"])["Automobile_Sales"].mean().reset_index()

plt.figure(figsize=(12,6))

sns.barplot(
    data=s,
    x="Vehicle_Type",
    y="Automobile_Sales",
    hue="Recession"
)

plt.title("Average Vehicle Sales During Recession and Non-Recession")
plt.xlabel("Vehicle Type")
plt.ylabel("Average Automobile Sales")

plt.xticks(rotation=45)
plt.yticks(rotation=45)

plt.show()
```
