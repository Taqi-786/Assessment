```python
#Ans1.
class ClinicAppointment:
    def __init__(self):
        self.appointments = []

    def book(self):
        name = input("Enter Name: ")
        age = input("Enter Age: ")
        mobile = input("Enter Mobile Number: ")
        doctor = input("Enter Doctor Name: ")

        print("Time Slots: 10am, 11am, 12pm, 2pm, 3pm")
        slot = input("Choose Time Slot: ")

        count = 0
        for a in self.appointments:
            if a["doctor"] == doctor and a["slot"] == slot:
                count += 1

        if count >= 3:
            print("Slot Full!")
            return

        self.appointments.append({
            "name": name,
            "age": age,
            "mobile": mobile,
            "doctor": doctor,
            "slot": slot
        })

        print("Appointment Booked Successfully!")

    def view(self):
        mobile = input("Enter Mobile Number: ")

        for a in self.appointments:
            if a["mobile"] == mobile:
                print(a)
                return

        print("Appointment Not Found!")

    def cancel(self):
        mobile = input("Enter Mobile Number: ")

        for a in self.appointments:
            if a["mobile"] == mobile:
                self.appointments.remove(a)
                print("Appointment Cancelled!")
                return

        print("Appointment Not Found!")

# Main Program
clinic = ClinicAppointment()

while True:
    print("\n1. Book Appointment")
    print("2. View Appointment")
    print("3. Cancel Appointment")
    print("4. Exit")

    choice = input("Enter Choice: ")

    if choice == "1":
        clinic.book()
    elif choice == "2":
        clinic.view()
    elif choice == "3":
        clinic.cancel()
    elif choice == "4":
        break
    else:
        print("Invalid Choice")
```

```python
#Ans2.
class SchoolManagement:
    def __init__(self):
        self.students = {}
        self.student_id = 1

    def admission(self):
        name = input("Enter Student Name: ")
        age = int(input("Enter Age: "))

        if age < 5 or age > 18:
            print("Age must be between 5 and 18")
            return

        std = input("Enter Class (1-12): ")
        mobile = input("Enter Guardian Mobile Number: ")

        if len(mobile) != 10 or not mobile.isdigit():
            print("Invalid Mobile Number")
            return

        self.students[self.student_id] = {
            "Name": name,
            "Age": age,
            "Class": std,
            "Mobile": mobile
        }

        print("Admission Successful")
        print("Student ID is:", self.student_id)

        self.student_id += 1

    def view(self):
        sid = int(input("Enter Student ID: "))

        if sid in self.students:
            print(self.students[sid])
        else:
            print("Student Not Found")

    def update(self):
        sid = int(input("Enter Student ID: "))

        if sid in self.students:
            new_class = input("Enter New Class: ")
            new_mobile = input("Enter New Mobile Number: ")

            if len(new_mobile) == 10 and new_mobile.isdigit():
                self.students[sid]["Class"] = new_class
                self.students[sid]["Mobile"] = new_mobile
                print("Record Updated")
            else:
                print("Invalid Mobile Number")
        else:
            print("Student Not Found")

    def remove(self):
        sid = int(input("Enter Student ID: "))

        if sid in self.students:
            del self.students[sid]
            print("Student Record Removed")
        else:
            print("Student Not Found")


# Main Program
school = SchoolManagement()

while True:
    print("\n1. New Admission")
    print("2. View Student Details")
    print("3. Update Student Info")
    print("4. Remove Student Record")
    print("5. Exit")

    choice = input("Enter Your Choice: ")

    if choice == "1":
        school.admission()
    elif choice == "2":
        school.view()
    elif choice == "3":
        school.update()
    elif choice == "4":
        school.remove()
    elif choice == "5":
        print("Thank You!")
        break
    else:
        print("Invalid Choice")
```

```python
#Ans3.
class BusReservation:
    def __init__(self):
        self.routes = {
            "1": ("Mumbai to Pune", 500),
            "2": ("Delhi to Jaipur", 600),
            "3": ("Surat to Ahmedabad", 300)
        }

        self.tickets = {}
        self.ticket_id = 1

    def show_routes(self):
        print("\nAvailable Routes")
        for key, value in self.routes.items():
            print(key, ".", value[0], "- ₹", value[1])

    def book_ticket(self):
        name = input("Enter Passenger Name: ")
        age = input("Enter Age: ")
        mobile = input("Enter Mobile Number: ")

        self.show_routes()
        choice = input("Choose Route: ")

        if choice not in self.routes:
            print("Invalid Route")
            return

        route = self.routes[choice][0]
        price = self.routes[choice][1]

        seat = 1
        for t in self.tickets.values():
            if t["Route"] == route:
                seat += 1

        if seat > 40:
            print("Bus is Full")
            return

        self.tickets[self.ticket_id] = {
            "Name": name,
            "Age": age,
            "Mobile": mobile,
            "Route": route,
            "Price": price,
            "Seat": seat
        }

        print("Ticket Booked Successfully")
        print("Ticket ID:", self.ticket_id)
        print("Seat Number:", seat)

        self.ticket_id += 1

    def view_ticket(self):
        tid = int(input("Enter Ticket ID: "))

        if tid in self.tickets:
            print(self.tickets[tid])
        else:
            print("Ticket Not Found")

    def cancel_ticket(self):
        tid = int(input("Enter Ticket ID: "))

        if tid in self.tickets:
            del self.tickets[tid]
            print("Ticket Cancelled")
        else:
            print("Ticket Not Found")


# Main Program
bus = BusReservation()

while True:
    print("\n===== Bus Reservation System =====")
    print("1. Show Available Routes")
    print("2. Book Ticket")
    print("3. View Ticket")
    print("4. Cancel Ticket")
    print("5. Exit")

    choice = input("Enter Your Choice: ")

    if choice == "1":
        bus.show_routes()
    elif choice == "2":
        bus.book_ticket()
    elif choice == "3":
        bus.view_ticket()
    elif choice == "4":
        bus.cancel_ticket()
    elif choice == "5":
        print("Thank You!")
        break
    else:
        print("Invalid Choice")
```
